```python
import torch
import sys
import os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
from model.encoder import Encoder
from model.decoder import Decoder
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

```python
# Hyperparameters (example)
vocab_size = 1000
context_length = 20    # max sequence length
model_dim = 64
num_blocks = 2
num_heads = 4
batch_size = 2
encoder_seq_len = 15
decoder_seq_len = 10
```

```python
# Instantiate encoder and decoder
encoder = Encoder(vocab_size, context_length, model_dim, num_blocks, num_heads).to(device)
decoder = Decoder(vocab_size, context_length, model_dim, num_blocks, num_heads).to(device)
```

```python
# Dummy inputs: random token IDs (batch_size x seq_len)
encoder_input_ids = torch.randint(0, vocab_size, (batch_size, encoder_seq_len)).to(device)
decoder_input_ids = torch.randint(0, vocab_size, (batch_size, decoder_seq_len)).to(device)
```

```python
# Forward pass
# encoder_output = encoder(encoder_input_ids)
encoder_hidden_states, encoder_logits = encoder(encoder_input_ids)
print("encoder_hidden_states.shape:", encoder_hidden_states.shape)  # Should be (batch_size, encoder_seq_len, model_dim)
print("encoder_logits.shape:", encoder_logits.shape)  # (batch_size, encoder_seq_len, vocab_size)
# print(f"Encoder output shape: {encoder_output.shape}")  
# Expected: (batch_size, encoder_seq_len, model_dim)

decoder_logits = decoder(decoder_input_ids, encoder_hidden_states)
print(f"Decoder logits shape: {decoder_logits.shape}")  
# Expected: (batch_size, decoder_seq_len, vocab_size)

# Check logits sum to 1 after softmax on a sample token (optional)
probs = torch.softmax(decoder_logits, dim=-1)
print(f"Sum of probs (first batch, first token): {probs[0,0].sum().item():.4f}")  # Should be ~1.0
```
